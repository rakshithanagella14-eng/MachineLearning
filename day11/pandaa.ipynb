Grouping by single column

```python
import pandas as pd
df = pd.read_csv("https://media.geeksforgeeks.org/wp-content/uploads/nba.csv")

team = df.groupby('Team')
print(team.first())
```

Grouping by multiple column

```python
import pandas as pd
df = pd.read_csv("https://media.geeksforgeeks.org/wp-content/uploads/nba.csv")

grouping = df.groupby(['Team', 'Position'])
print(grouping.first())
```

Applying Aggregation by GroupBy

```python
import pandas as pd
df = pd.read_csv("https://media.geeksforgeeks.org/wp-content/uploads/nba.csv")

aggregated_data = df.groupby(['Team', 'Position']).agg(
   total_salary=('Salary', 'sum'),
   avg_salary=('Salary', 'mean'),
   player_count=('Name', 'count')
)

print(aggregated_data)
```

Applying transformations method

```python
import pandas as pd
df = pd.read_csv("https://media.geeksforgeeks.org/wp-content/uploads/nba.csv")

df['Rank within Team'] = df.groupby('Team')['Salary'].transform(lambda x: x.rank(ascending=False))
print(df)
```

Filtering groups

```python
import pandas as pd
df = pd.read_csv("https://media.geeksforgeeks.org/wp-content/uploads/nba.csv")

filtered_df = df.groupby('Team').filter(lambda x: x['Salary'].mean() >= 1000000)
print(filtered_df)
```
