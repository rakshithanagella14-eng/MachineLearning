Merging

```python
import pandas as pd

data1 = [[1, 'Raj', 25], 
         [2, 'Sharma', 42], 
         [3, 'Saroj', 22], 
         [4, 'Raja', 51], 
         [5, 'Kajal', 26]]
df1 = pd.DataFrame(data1, columns=['ID', 'Name', 'Age'])

data2 = [[1, 'Male', 25000], 
         [2, 'Male', 42500], 
         [3, 'Female', 22300], 
         [4, 'Male', 51400], 
         [5, 'Female', 26800]]
df2 = pd.DataFrame(data2, columns=['ID', 'Gender', 'Salary'])

print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)
```

Inner merge

```python
merged_df = pd.merge(df1, df2, on='ID', how='inner')
print(merged_df)
```

Left merge

```python
merged_df = pd.merge(df1, df2, on='ID', how='left')
print(merged_df)
```

Right merge

```python
merged_df = pd.merge(df1, df2, on='ID', how='right')
print(merged_df)
```

Outer merge

```python
merged_df = pd.merge(df1, df2, on='ID', how='outer')
print(merged_df)
```

Merging on multiple columns

```python
df1['Country'] = ['USA', 'Canada', 'USA', 'Canada', 'USA']
df2['Country'] = ['USA', 'Canada', 'USA', 'Canada', 'Mexico']

merged_df = pd.merge(df1, df2, on=['ID', 'Country'], how='inner')
print(merged_df)
```

merging with different column names

```python
df2.rename(columns={'ID': 'EmployeeID'}, inplace=True)

merged_df = pd.merge(df1, df2, left_on='ID', right_on='EmployeeID', how='inner')
print(merged_df)
```

creating sample dataframe

```python
import pandas as pd

# creating a dataframe
df = pd.DataFrame({'A': ['John', 'Boby', 'Mina'],
                   'B': ['Masters', 'Graduate', 'Graduate'],
                   'C': [27, 23, 21]})

df
```

creating and pivot a dataframe

```python
df.pivot_table(index='A', columns='B', values='C', aggfunc='mean')
```

multilevel pivot table with Pandas DataFrame

```python
df.pivot(index='A', columns='B', values=['C', 'A'])
```
